```python
from ultralytics import YOLO
import cv2
import os

# loading the pretrained YOLOv8 model nano
model = YOLO("/kaggle/working/yolov8n.pt")  

# video path for Jurassic Park scene
input_video_path = "/kaggle/input/jurasic-world-movie-scene/movie_scene_jurasic_park.mp4"
output_video_path = "/kaggle/working/raptor_chase_detected.mp4"

# opening the video file
cap = cv2.VideoCapture(input_video_path)

# getting video properties
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_video_path, fourcc, fps, (frame_width, frame_height))

# processing the video frame by frame
# and detecting objects using YOLOv8
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    results = model(frame)

    # annotating the frame with detected objects
    annotated_frame = results[0].plot()

    # writing annotated frame to output
    out.write(annotated_frame)

# releasing video capture and writer objects
cap.release()
out.release()

print("✅ Detection done! Output saved as:", output_video_path)
```
